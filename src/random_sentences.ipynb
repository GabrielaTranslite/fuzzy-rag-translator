```python
import pandas as pd
from retrieve import load_translation_memory
```

```python
# Importing the translation memory
tm = load_translation_memory("../data/tm/translation_memory.jsonl")
```

```python
tm_df = pd.DataFrame(tm)
tm_df.head(10)
```

```python
sample_set = tm_df[(tm_df["source_norm"].str.len() > 20) & (tm_df["source_norm"].str.len() < 200)]
sample_sentences = sample_set.sample(10, random_state=42)
pd.set_option("display.max_colwidth", 200)
chosen_sample = sample_sentences["source_norm"]
print(chosen_sample)
```

```python
chosen_sample.to_excel("../data/random_sentences.xlsx", index=False)
```
