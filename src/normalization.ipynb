```python
import json
import re

```

```python
# Importing gold set to list of dictionaries
with open("../eval/gold_set.jsonl", "r", encoding="utf-8") as file:
    gold_set = [json.loads(row) for row in file if row.strip()]
```

```python
pattern = r'\^[A-Za-z]'
for record in gold_set:
    if re.search(pattern, record['query']):
        print(record['query'])
```

```python
for record in gold_set:
    query = record.get("query", "")
    if "^" in query:
        dev_comment, _, new_query = query.partition("^")
        record["dev_comment"] = dev_comment.strip()
        record["query"] = new_query.strip()
```

```python
# Testing the solution
for record in gold_set:
    if record.get("dev_comment"):
        dev_comment = record["dev_comment"]
        clean_query = record["query"]
        print(f"Query: {clean_query}, developer's comment: {dev_comment}")
        
```
