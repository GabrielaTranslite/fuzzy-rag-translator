## Fuzzy Matching

```python
import json
```

```python
# Importing TM as a list

with open("data/tm/translation_memory.jsonl", "r", encoding="utf-8") as f:
    source = f.readlines()
translation_memory = [json.loads(line) for line in source]
print(f"Loaded {len(translation_memory)} translation memory entries.")
```

```python
print(translation_memory[0])  # Print the first entry to verify the structure
```

```python
from rapidfuzz import process, fuzz

ratio =fuzz.ratio(translation_memory[0]["source"], translation_memory[1]["source"])
print(f"Fuzzy match ratio between first two entries: {ratio}")
```

```python
test_string = "Click on Victoria"

top_entries = sorted(translation_memory, key=lambda x: fuzz.ratio(test_string, x["source"]), reverse=True)[:3]
print(f"Top 3 fuzzy matches for '{test_string}':")
for match in top_entries:
    print(f"Source: {match['source']}, Target: {match['target']}, Score: {round(fuzz.ratio(test_string, match['source']), 2)}")
```

```python
# Building the list of source strings to reuse it for every query

sources = [entry["source"] for entry in translation_memory]

matches = process.extract(test_string, sources, scorer=fuzz.ratio, limit=3)

print(f"Top 3 matches using process.extract for '{test_string}':")
for matched_source, score, index in matches:
    record = translation_memory[index]
    print(f"{score:.1f}  {record['source']}  ->  {record['target']}")
```
